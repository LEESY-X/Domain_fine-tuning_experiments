# PAPER 실행 진행상태 모니터

학습 노트북과 다른 커널에서 실행하면 학습 중에도 상태를 확인할 수 있습니다.


```python
from pathlib import Path
import json
from collections import Counter
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
```

```python
STUDY = 'study1'  # study1, study2, study3
RUN_MODE = 'PAPER'
```

```python
root = ROOT / 'results' / STUDY / RUN_MODE
rows = []
for path in root.glob('**/status.json') if root.exists() else []:
    try:
        row = json.loads(path.read_text(encoding='utf-8')); row['path'] = str(path.parent.relative_to(ROOT)); rows.append(row)
    except Exception:
        rows.append({'status':'UNREADABLE','path':str(path.parent.relative_to(ROOT))})
print('STATUS', dict(Counter(x.get('status','UNKNOWN') for x in rows)))
for x in rows:
    if x.get('status') in {'RUNNING','FAILED'}: print(x.get('status'),x.get('task'),x.get('model'),x.get('method'),x.get('seed'),x.get('resumed_from_checkpoint'))
progress = root / 'progress.json'
print(progress.read_text(encoding='utf-8') if progress.exists() else 'progress.json 없음')
```
